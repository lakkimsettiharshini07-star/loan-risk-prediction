```python
#  Keep only salaried-type applicants and test the DAYS_EMPLOYED placeholder.

# The income types we treat as salaried (assumption: Commercial associate
# means private-sector employee. The dataset does not define it, so we
# state this assumption in the README).
salaried_types = ["Working", "State servant", "Commercial associate"]

# df["NAME_INCOME_TYPE"].isin(...) gives True/False for each row.
# df[ True/False list ] keeps only the True rows (this is called a boolean mask).
# .copy() makes an independent table so later edits don't touch the original df.
df_sal = df[df["NAME_INCOME_TYPE"].isin(salaried_types)].copy()

print("Rows before filter:", len(df))
print("Rows after filter: ", len(df_sal))

# Target rate in the filtered data (mean of a 0/1 column = share of 1s)
print("Share of TARGET = 1 after filter (%):", round(df_sal["TARGET"].mean() * 100, 2))

# The placeholder value. A normal DAYS_EMPLOYED is negative; 365243 is not real.
PLACEHOLDER = 365243

# (df["DAYS_EMPLOYED"] == PLACEHOLDER) gives True/False per row; .sum() counts the Trues.
print("Placeholder rows in full data:    ", (df["DAYS_EMPLOYED"] == PLACEHOLDER).sum())
print("Placeholder rows in salaried data:", (df_sal["DAYS_EMPLOYED"] == PLACEHOLDER).sum())

# Which income types hold the placeholder? (tests our suspicion)
print()
print(df.loc[df["DAYS_EMPLOYED"] == PLACEHOLDER, "NAME_INCOME_TYPE"].value_counts())

# Missing occupations after filtering
missing_occ = df_sal["OCCUPATION_TYPE"].isnull().sum()
print()
print("Missing OCCUPATION_TYPE in salaried data:", missing_occ,
      f"({missing_occ / len(df_sal) * 100:.1f}%)")
```

```python
# Count invalid values: amounts that are zero or negative cannot be real
for col in ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "AMT_GOODS_PRICE"]:
    # (condition).sum() counts the rows where the condition is True
    # (missing values are not counted, because NaN <= 0 is False)
    print(col, "<= 0:", (df_sal[col] <= 0).sum())

# DAYS_EMPLOYED should be 0 or negative (days BEFORE the application)
print("DAYS_EMPLOYED > 0:", (df_sal["DAYS_EMPLOYED"] > 0).sum())
```

```python
# How many applicants exceed income cutoffs at different percentiles?
# quantile(q) returns the value below which a fraction q of rows fall.
# (condition).sum() counts rows where the condition is True.

for q in [0.99, 0.999, 0.9999]:
    cutoff = df_sal["AMT_INCOME_TOTAL"].quantile(q)
    n_above = (df_sal["AMT_INCOME_TOTAL"] > cutoff).sum()
    print(f"{q * 100:.2f}th percentile cutoff = {cutoff:,.0f} -> rows above: {n_above}")
```

```python
# Default rate (share of TARGET = 1) for each category.
# Why: shows which groups look riskier, and which groups are too small to trust.

# An earlier cell set one decimal place for display; use two decimals here
pd.options.display.float_format = "{:,.2f}".format

# Fill missing occupations with "Unknown" ONLY for this view,
# so that those rows appear as their own group
occupation = df_sal["OCCUPATION_TYPE"].fillna("Unknown")

groups = {
    "NAME_CONTRACT_TYPE": df_sal["NAME_CONTRACT_TYPE"],
    "NAME_INCOME_TYPE": df_sal["NAME_INCOME_TYPE"],
    "NAME_EDUCATION_TYPE": df_sal["NAME_EDUCATION_TYPE"],
    "OCCUPATION_TYPE": occupation,
}

for name, column in groups.items():
    # groupby splits the rows into one group per category.
    # agg computes two numbers for each group:
    #   applicants   = how many rows are in the group (count)
    #   default_rate = mean of a 0/1 column = share of 1s
    summary = (
        df_sal.groupby(column)["TARGET"]
        .agg(applicants="count", default_rate="mean")
        .sort_values("default_rate", ascending=False)
    )
    summary["default_rate"] = summary["default_rate"] * 100   # as percentage
    print(name)
    print(summary)
    print()
```

```python
# Default rate by occupation, printed in full (not shortened)
# Missing occupations are shown as their own group called "Unknown"
occupation = df_sal["OCCUPATION_TYPE"].fillna("Unknown")

summary = (
    df_sal.groupby(occupation)["TARGET"]
    .agg(applicants="count", default_rate="mean")
    .sort_values("default_rate", ascending=False)
)
summary["default_rate"] = summary["default_rate"] * 100

# to_string() prints every row, so nothing is hidden
print(summary.to_string())
```

```python
# Preview two ratio features and check whether they relate to default rate.
# We build them in a separate small table, so df_sal is not changed.

ratios = pd.DataFrame({
    # loan size relative to annual income
    "loan_to_income": df_sal["AMT_CREDIT"] / df_sal["AMT_INCOME_TOTAL"],
    # annuity (payment) relative to annual income
    "annuity_to_income": df_sal["AMT_ANNUITY"] / df_sal["AMT_INCOME_TOTAL"],
    "TARGET": df_sal["TARGET"],
})

# Summary statistics: look at the typical values and the extreme ends
pd.options.display.float_format = "{:,.3f}".format
print(ratios[["loan_to_income", "annuity_to_income"]]
      .describe(percentiles=[0.01, 0.5, 0.99]).T)

# Default rate by ratio band.
# pd.qcut splits rows into 5 equal-sized groups (quintiles) by ratio value.
# duplicates="drop" avoids an error if two band edges are equal.
for col in ["loan_to_income", "annuity_to_income"]:
    bands = pd.qcut(ratios[col], q=5, duplicates="drop")
    # observed=True keeps only bands that actually occur
    rate = ratios.groupby(bands, observed=True)["TARGET"].mean() * 100
    print()
    print(col, "- default rate (%) by band, lowest to highest ratio")
    print(rate.round(2))
```

```python
# Compare numeric features on one scale: how much does the default rate
# change from the lowest group to the highest group of each feature?

feat = pd.DataFrame({
    "AMT_INCOME_TOTAL": df_sal["AMT_INCOME_TOTAL"],
    "AMT_CREDIT": df_sal["AMT_CREDIT"],
    "AMT_ANNUITY": df_sal["AMT_ANNUITY"],
    "AMT_GOODS_PRICE": df_sal["AMT_GOODS_PRICE"],
    # DAYS_EMPLOYED is negative days before the application; convert to years
    "employment_years": -df_sal["DAYS_EMPLOYED"] / 365.25,
    "loan_to_income": df_sal["AMT_CREDIT"] / df_sal["AMT_INCOME_TOTAL"],
    "annuity_to_income": df_sal["AMT_ANNUITY"] / df_sal["AMT_INCOME_TOTAL"],
    "TARGET": df_sal["TARGET"],
})

rows = []
for col in feat.columns.drop("TARGET"):          # every column except TARGET
    bands = pd.qcut(feat[col], q=5, duplicates="drop")   # 5 equal-sized groups
    rate = feat.groupby(bands, observed=True)["TARGET"].mean() * 100
    rows.append({
        "feature": col,
        "lowest_band_rate": rate.iloc[0],        # first group (smallest values)
        "highest_band_rate": rate.iloc[-1],      # last group (largest values)
        "spread_pp": rate.max() - rate.min(),    # gap in percentage points
    })

spread_table = pd.DataFrame(rows).set_index("feature")

# Spearman correlation: do higher feature values go with TARGET = 1?
# +1 = always, -1 = never, 0 = no consistent link. It uses ranks, so
# extreme values such as the 117,000,000 income do not distort it.
spread_table["spearman_with_target"] = feat.corr(method="spearman")["TARGET"].drop("TARGET")

print(spread_table.round(3).sort_values("spread_pp", ascending=False))
```

```python
# Leakage and redundancy checks on the candidate features

# 1) An ID should carry no information about the outcome.
#    A Spearman value near 0 is what we expect (we will not use the ID anyway).
print("SK_ID_CURR vs TARGET (Spearman):",
      round(df_sal["SK_ID_CURR"].corr(df_sal["TARGET"], method="spearman"), 4))

# 2) Redundancy: do credit and goods price carry nearly the same information?
#    dropna() removes rows where either value is missing.
both = df_sal[["AMT_CREDIT", "AMT_GOODS_PRICE"]].dropna()
print("Credit vs goods price (Spearman):",
      round(both["AMT_CREDIT"].corr(both["AMT_GOODS_PRICE"], method="spearman"), 3))
# Comparing two columns gives True/False per row; mean() of that is the share of True
print("Rows where credit equals goods price (%):",
      round((both["AMT_CREDIT"] == both["AMT_GOODS_PRICE"]).mean() * 100, 1))

# 3) Redundancy among the other numeric features.
#    Values close to +1 or -1 mean two features say nearly the same thing.
print()
print(feat.drop(columns=["AMT_GOODS_PRICE", "TARGET"])
          .corr(method="spearman").round(2))
```

## Exploration findings
- Salaried filter (Working, State servant, Commercial associate): 252,094 rows, 8.66% TARGET = 1.
- No zero or negative amounts. DAYS_EMPLOYED placeholder (365243) occurs only for Pensioner/Unemployed, so none remain.
- AMT_INCOME_TOTAL has extreme outliers (max 117,000,000 vs 99.9th percentile 945,000).
  Plan: cap at the 99.9th percentile, calculated on training data only.
- Default rate varies most by occupation and education; among numeric features, employment_years.
- loan_to_income is non-monotonic against default rate; annuity_to_income rises slightly.
- AMT_GOODS_PRICE is nearly redundant with AMT_CREDIT (Spearman 0.984) and is not collected on the form: dropped.
- AMT_ANNUITY may not be a monthly payment (median annuity is about twice median monthly income): mapping from the form's monthly instalment is an assumption.
- No single feature is strongly predictive (|Spearman| <= 0.08), so a modest model is expected.
- Assumption: "Commercial associate" is private-sector salaried (not defined in the dataset).
- Fairness note: education and occupation may act as proxies for background.
